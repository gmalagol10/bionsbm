```python
for dataset, job in zip(
    ["PBMC", "HSPC", "MouseSkin", "BMMCMultiOme"],
    ["Pb", "Hs", "Ms", "Bm"],
):
    for exp, n in zip(
        ["Peak_GEX", "Peak_mRNA", "Peak_lncRNA", "Peak_mRNA_lncRNA", "mRNA_lncRNA"],
        ["PH", "PM", "PL", "PML", "ML"],
    ):
        cm = f"Datasets/{dataset}/CM/{dataset}_{exp}_Def.h5mu"
        path_to_save = (
            f"Datasets/{dataset}/COBOLT/COBOLT_{exp}/"
            f"{dataset}_COBOLT_{exp}"
        )

        os.system(
            f"sbatch -p slim16 -J Cob{job}{n} "
            f"--time=7-00:00:00 --mem=32G --ntasks=1 --cpus-per-task=8 "
            f"--output=Datasets/{dataset}/{dataset}_{exp}_COBOLT.out "
            f"Run_Exp.sh COBOLT {cm} {path_to_save} 0"
        )
```

```python
import subprocess
from pathlib import Path

SUBMIT = False
N_RUNS = 25

jobs =[
    (dataset, exp, f"Cob{job}{n}", "slim16")
    for dataset, job in zip(
        ["PBMC", "HSPC", "MouseSkin", "BMMCMultiOme"],
        ["Pb", "Hs", "Ms", "Bm"],
    )
    for exp, n in zip(
        ["Peak_GEX", "Peak_mRNA", "Peak_lncRNA", "Peak_mRNA_lncRNA", "mRNA_lncRNA"],
        ["PH", "PM", "PL", "PML", "ML"],
    )
]

for dataset, exp, job, partition in jobs:
    cm = f"Datasets/{dataset}/CM/{dataset}_{exp}_Def.h5mu"
    prefix = Path(
        f"Datasets/{dataset}/COBOLT/COBOLT_{exp}/"
        f"{dataset}_COBOLT_{exp}"
    )

    start = next(
        (
            run
            for run in range(N_RUNS)
            if not (
                prefix.parent
                / "Runs"
                / f"Run{run}"
                / f"{prefix.name}_Embedding.tsv.gz"
            ).is_file()
        ),
        N_RUNS,
    )

    if start == N_RUNS:
        continue

    print(dataset, exp, start)

    if SUBMIT:
        subprocess.run(
            [
                "sbatch",
                "-p", partition,
                "-J", f"{job}r{start}",
                "--time=7-00:00:00",
                "--mem=32G",
                "--ntasks=1",
                "--cpus-per-task=8",
                f"--output=Datasets/{dataset}/{dataset}_{exp}_COBOLT.out",
                "Run_Exp.sh",
                "COBOLT",
                cm,
                str(prefix),
                str(start),
            ],
            stdout=subprocess.DEVNULL,
            check=True,
        )
```
